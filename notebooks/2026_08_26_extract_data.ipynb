```python
import pandas as pd
import somadata as sd
```

```python
adat = sd.read_adat(
    "/mnt/data/raw/SS-217041.raw.adat"
)
```

```python
# Extract Feature Data
features = adat.columns.to_frame(index=False).rename({"SeqId": "ProbeId"}, axis = 1)
features.to_csv(
    "/mnt/data/processed/features.csv",
    index=False
)
features
```

```python
# Extract Sample Data
samples = adat.index.to_frame().reset_index(drop = True)
samples.to_csv(
    "/mnt/data/processed/samples.csv",
    index=False
)
samples
```

```python
# Extract Measurement Data
measurements = adat.reset_index(drop = True)
measurements.columns = features["ProbeId"]
measurements = pd.concat(
    [
        samples.loc[:, ["PlateId", "PlatePosition"]],
        measurements
    ],
    axis=1
)
measurements = measurements.melt(
    id_vars=["PlateId", "PlatePosition"],
    var_name="ProbeId"
)
measurements.to_csv(
    "/mnt/data/processed/measurements.raw.csv",
    index=False
)
measurements
```
